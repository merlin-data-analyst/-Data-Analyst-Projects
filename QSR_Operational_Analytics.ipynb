```python
#ROOT CAUSE ANALYSIS(1)-Investigate Staff Efficiency(Create efficiency groups)
df_clean['efficiency_category'] = pd.cut(
    df_clean['staff_efficiency'],
    bins=[0, 7, 8, 9, 10],
    labels=['Low', 'Medium', 'High', 'Very High'],
    include_lowest=True
)
```

```python
efficiency_analysis = (
    df_clean.groupby('efficiency_category', observed=True)
    .agg(
        total_orders=('order_id', 'count'),
        avg_wait_time=('wait_time_min', 'mean'),
        avg_order_duration=('order_duration_min', 'mean'),
        avg_satisfaction=('customer_satisfaction', 'mean')
    )
    .round(2)
    .reset_index()
)

efficiency_analysis
```

```python
#ROOT CAUSE ANALYSIS (2) (low-efficiency vs high-efficiency orders.)
efficiency_median = df_clean['staff_efficiency'].median()

efficiency_comparison = pd.DataFrame({
    'Group': [
        'Below median efficiency',
        'At/above median efficiency'
    ],
    
    'Orders': [
        (df_clean['staff_efficiency'] < efficiency_median).sum(),
        (df_clean['staff_efficiency'] >= efficiency_median).sum()
    ],
    
    'Average Wait Time': [
        df_clean.loc[
            df_clean['staff_efficiency'] < efficiency_median,
            'wait_time_min'
        ].mean(),
        
        df_clean.loc[
            df_clean['staff_efficiency'] >= efficiency_median,
            'wait_time_min'
        ].mean()
    ],
    
    'Average Satisfaction': [
        df_clean.loc[
            df_clean['staff_efficiency'] < efficiency_median,
            'customer_satisfaction'
        ].mean(),
        
        df_clean.loc[
            df_clean['staff_efficiency'] >= efficiency_median,
            'customer_satisfaction'
        ].mean()
    ]
}).round(2)

efficiency_comparison
```

```python
#ROOT CAUSE ANALYSIS (3) - low staff efficiency creates long waits.
efficiency_long_wait = (
    df_clean.assign(
        long_wait=df_clean['wait_time_min'] > 10
    )
    .groupby('efficiency_category', observed=True)
    .agg(
        total_orders=('order_id', 'count'),
        long_wait_orders=('long_wait', 'sum'),
        avg_wait_time=('wait_time_min', 'mean'),
        avg_satisfaction=('customer_satisfaction', 'mean')
    )
    .reset_index()
)

efficiency_long_wait['long_wait_percentage'] = (
    efficiency_long_wait['long_wait_orders']
    / efficiency_long_wait['total_orders'] * 100
).round(2)

efficiency_long_wait

#We're asking? ans:Does lower staff efficiency produce a higher percentage of long-wait orders?
```
